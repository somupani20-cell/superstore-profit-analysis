```python
tables = df[df["Sub-Category"] == "Tables"]
print(tables["Discount"].mean())
print(df["Discount"].mean())
```

```python
tables_light = tables[tables["Discount"] < 0.3]
tables_heavy = tables[tables["Discount"] >= 0.3]
print(tables_light["Profit"].sum())
print(tables_heavy["Profit"].sum())
```
